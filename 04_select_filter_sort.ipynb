**SET UP NOTEBOOK**

```python
import pandas as pd

df = pd.read_csv("train.csv")

cols = ["SalePrice", "GrLivArea", "OverallQual", "YearBuilt", "Neighborhood", "BedroomAbvGr"]
small = df[cols].copy()
small.head()
```

**SELECT COLUMNS**

```python
small["SalePrice"]   # one column
small[["SalePrice", "GrLivArea"]]  # two columns
small.loc[0]                       # one row, the first house
small.loc[0:2, ["SalePrice", "YearBuilt"]]  # rows 0 to 2, two columns
```

```python
small[["Neighborhood", "YearBuilt"]].head()
```

**FILTER ROWS**

```python
big_homes = small[small["GrLivArea"] > 2000]
print(len(big_homes))
big_homes.head()
```

```python
small[small["Neighborhood"] == "NoRidge"].head()
```

```python
good_new = small[(small["YearBuilt"]  > 2000) & (small["OverallQual"] >= 8)]
print(len(good_new))
good_new.head()
```

```python
luxury_homes = small[(small["SalePrice"] > 400000) & (small["BedroomAbvGr"] >= 4)]
print(len(luxury_homes))
luxury_homes.head()
```

**5 MOST EXPENSIVE**

```python
small.sort_values("SalePrice", ascending = False).head(5)
```

**THE 5  CHEAPEST HOUSES**

```python
small.sort_values("SalePrice")
```

**SORT BY TWO COLUMNS**

```python
small.sort_values(["OverallQual", "SalePrice"], ascending = [False, False]).head(5)
```
