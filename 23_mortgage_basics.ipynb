**HOW A MORTGAGE CHANGES THE RETURNS**

```python
price = 590000
deposit_pct = 0.25
rate = 0.05       # example yearly interest rate
years = 25        # example loan term

loan = price * (1 - deposit_pct)
r = rate / 12
n = years * 12
payment = loan * r / (1 - (1 + r) ** -n)

noi_year =  18743 # from my Didcot analysis (service charge yearly case)
cash_flow_month = noi_year / 12 - payment

print("Deposit:", round(price * deposit_pct))
print("Loan:", round(loan))
print("Monthly payment:", round(payment))
print("Total interest over the loan:", round(payment * n - loan))
print("Monthly income before mortgage (NOI / 12):", round(noi_year / 12))
print("Monthly cash flow after mortgage:", round(cash_flow_month))
```

```python
import pandas as pd

rows = []
for rate in [0.03, 0.04, 0.05, 0.06, 0.07]:
  r = rate / 12
  pay = loan * r / (1 - (1 + r) ** -n)
  rows.append({"Rate %": rate * 100,
               "Monthly payment": round(pay),
               "Monthly cash flow": round(noi_year / 12 - pay)})
pd.DataFrame(rows)
```

*With a 25% deposit (£147,500) and a 5% mortgage over 25 years, the monthly payment is about £2,587, against about £1,562 a month of income after costs, so the house loses about £1,025 a month. Even at 3% the cash flow is negative (about £536 a month). Each extra percentage point costs about £250 more a month, and the 25-year interest total is about £333,500. As a pure rental, the house does not pay for itself at these assumptions, and any case for buying would depend on the price rising. These are example assumptions, not advice, and they leave out stamp duty, tax, and mortgage fees.*
