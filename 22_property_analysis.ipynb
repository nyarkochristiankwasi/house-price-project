**Analyzing a real listing, Cala Plot 503, Didcot.**

### Assumptions
Price £590,000. Rent £2,500 a month (middle of six comparable 4-bed detached rentals, range £2,000 to £2,750). Service charge £206.65 (period unconfirmed, so both versions are tested). Insurance about £350 a year, repairs 1% of price, vacancy 6% of rent, letting agent 10% of rent. All figures before mortgage and tax.

```python
import pandas as pd

price = 590000
insurance = 350
repairs = 0.01 * price
void_rate = 0.06
agent_rate = 0.10
service = 206.65

def analyze(rent_month, service_per_year):
  rent_year = rent_month * 12
  costs = (void_rate * rent_year + agent_rate * rent_year
           + insurance + repairs + service_per_year)
  noi = rent_year - costs
  return {
      "Monthly rent": rent_month,
      "Service charge": "yearly" if service_per_year < 1000 else "monthly",
      "Gross yield %": round(rent_year / price * 100, 2),
      "Costs per year": round(costs),
      "NOI per year": round(noi),
      "Cap rate %": round(noi / price * 100, 2),
  }


rows = []
for rent in [2000, 2500, 2750]:
  rows.append(analyze(rent, service)) # service charge is yearly
  rows.append(analyze(rent, service * 12))  #service charge is monthly

pd.DataFrame(rows)
```

*At £2,500 a month, Plot 503 has a gross yield of about 5.1% but a cap rate of only about 3.2% (2.8% if the service charge is monthly), because costs take about 37.5% of the rent. The rent comparables (£2,000 to £2,750) move the cap rate from 2.3% to 3.6%, about three times as much as the service charge uncertainty (about 0.4 points). The largest cost is the repairs allowance (1% of price), which may be cautious for a new build. All figures are before mortgage, stamp duty, and tax.*
