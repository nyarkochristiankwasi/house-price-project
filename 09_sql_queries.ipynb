**SQLITE_3**

```python
import sqlite3
import pandas as pd

df = pd.read_csv("train.csv")
conn = sqlite3.connect("housing.db")
df.to_sql("houses", conn, if_exists = "replace", index = False)

def q(sql):
  return pd.read_sql(sql, conn)
```

```python
q("SELECT Neighborhood, SalePrice, YearBuilt FROM houses WHERE SalePrice > 300000")
```

```python
q("SELECT Neighborhood, SalePrice FROM houses WHERE Neighborhood = 'NoRidge' ORDER BY SalePrice DESC")
```

```python
q("SELECT Neighborhood, SalePrice, GrLivArea FROM houses WHERE GrLivArea > 3000 ORDER BY SalePrice")
```

*Location matters alongside size: even the cheapest NoRidge house (190,000) sold above the overall median (163,000), while the two largest houses, both in Edwards, were the cheapest of all houses over 3,000 sq ft.*

```python
q("SELECT OverallQual, SalePrice, Neighborhood, YearBuilt FROM houses WHERE YearBuilt < 1950 AND OverallQual >=7 ORDER BY SalePrice DESC ")
```

*Among the 46 houses built before 1950 with a quality score of 7 or higher, 26 sold above the overall median (163,000). Most (35) are in OldTown and Crawfor, and Crawfor’s typical price (about 230,000) is well above OldTown’s (about 158,000), so location matters alongside quality.*
