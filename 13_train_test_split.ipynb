**TRAIN AND TEST SPLIT**

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("train.csv")

train, test = train_test_split(df, test_size=0.2, random_state=42)

print("Training houses:", len(train))
print("Test houses:", len(test))
print("Average price, training:", round(train["SalePrice"].mean()))
print("Average price, test:", round(test["SalePrice"].mean()))
```

*The training and test averages (181,442 and 178,840) are close because the rows were shuffled before splitting, so each set is a random sample of the same data. They differ slightly because a random sample never matches exactly.*
