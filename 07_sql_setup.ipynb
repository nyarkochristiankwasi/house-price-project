**LOAD THE HOUSING DATA INTO SQLITE**

```python
import sqlite3
import pandas as pd

df = pd.read_csv("train.csv")

conn = sqlite3.connect("housing.db")
df.to_sql("houses", conn, if_exists="replace", index=False)
```

```python
pd.read_sql("SELECT COUNT(*) AS number_of_houses FROM houses", conn)
```

**LOOKING AT 5 HOUSES**

```python
# Looking at 5 houses
pd.read_sql("SELECT Neighborhood, SalePrice, GrLivArea FROM houses LIMIT 5", conn)
```

**THE 5 MOST EXPENSIVE HOUSES**

```python
pd.read_sql("SELECT Neighborhood, SalePrice, GrLivArea FROM houses ORDER BY SalePrice DESC LIMIT 5", conn)
```

**FILTERING IN SQL**

```python
pd.read_sql("SELECT Neighborhood, SalePrice, GrLivArea FROM houses WHERE SalePrice > 400000 AND BedroomAbvGr >= 4",conn)
```

**USING GROUPBY**

```python
pd.read_sql("""
SELECT Neighborhood, COUNT(*) AS houses
FROM houses
WHERE SalePrice > 400000 AND BedroomAbvGr >= 4
GROUP BY Neighborhood
ORDER BY houses DESC
""", conn)
```
