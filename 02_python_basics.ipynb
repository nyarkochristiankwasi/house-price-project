```python
price = 250000
bedrooms = 3
city = "Ames"
print(price, bedrooms, city)
print("Price per bedroom:", price/bedrooms)
```

```python
prices = [250000, 180000, 320000, 150000]

print(prices[0]) #first item (counting start at 0)
print(len(prices))  #how many times
prices.append(275000)   #add one
print(max(prices), min(prices))
print(sum(prices) / len(prices))  #the average
print(prices)
```

```python
bedroom_count = [3,5,2,7,5]
print(max(bedroom_count))
print(sum(bedroom_count) / len(bedroom_count))
```

**DICTIONARIES**

```python
house = {"price": 250000, "bedrooms": 3, "city": "Ames"}

print(house["price"])
house["bathrooms"] = 2
print(house)
```

```python
house_dict = {"price": 10000000, "kitchen": 4, "bedrooms": 8, "hall": 5}
house_dict["garage"] = 4
print(house_dict)
```

**LOOPS**

```python
prices = [250000, 180000, 320000, 150000]

for p in prices:
  print("House costs", p)

for p in prices:
  if p > 200000:
    print(p, "is expensive")
  else:
    print(p, "is affordable")
```

```python
bedroom_count = [3, 5, 2, 7, 5]

for b in bedroom_count:
  print("number of bedrooms:", b)

for b in bedroom_count:
  if b > 4:
    print(b, "is big")
  else:
    print(b, "is small")
```

```python
bedroom_count = [1,6,2,7,9]

big_count = 0       #New: start the counter before the loop

for beds in bedroom_count:
  if beds >= 4:
    print(beds, "is big")
    big_count = big_count + 1 #New: Inside the if, so it only counts big houses

  else:
    print(beds, "is small")

print("Big houses:", big_count)    #New: After the loop, with no indent
```
