**FUNCTIONS**

```python
def price_per_bedroom(price, bedrooms):
  return price / bedrooms

print(price_per_bedroom(250000, 3))
print(price_per_bedroom(180000, 2))
```

```python
def size_label(bedrooms):
  if bedrooms >= 4:
    return "big"
  else:
    return "small"

print(size_label(5))
print(size_label(2))
```

```python
def is_expensive(price):
  if price > 200000:
    return True
  else:
    return False

print(is_expensive(345876543))
print(is_expensive(3445))
```
