**SETTING UP AND FINDING MISSING VALUES**

```python
import pandas as pd

df = pd.read_csv("train.csv")

missing = df.isna().sum()
missing = missing[missing > 0].sort_values(ascending = False)
print(len(missing))
missing
```

```python
df["PoolQC"].isna().mean() * 100
```

**FILLING THE MISSING VALUES**

```python
no_feature_cols = ["PoolQC", "MiscFeature", "Alley", "Fence", "FireplaceQu",
                   "GarageType", "GarageFinish", "GarageQual", "GarageCond",
                   "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2"
                   ]

df[no_feature_cols] = df[no_feature_cols].fillna("None")
```

**FILLING THE MISSING VALUES WITH THE MEDIAN**

```python
df["LotFrontage"] = df["LotFrontage"].fillna(df["LotFrontage"].median())
```

```python
remaining = df.isna().sum()
remaining[remaining > 0]
```

**FILLING THE LAST FOUR COLUMNS**

```python
df["MasVnrType"] = df["MasVnrType"].fillna("None")
df["MasVnrArea"] = df["MasVnrArea"].fillna(0)
df["Electrical"] = df["Electrical"].fillna(df["Electrical"].mode()[0])
df["GarageYrBlt"] = df["GarageYrBlt"].fillna(0)

print(df.isna().sum().sum())
```

**AVERAGE PRICE BY NEIGHBORHOOD**

```python
by_hood = df.groupby("Neighborhood")["SalePrice"].agg(["count", "mean", "median"])
by_hood = by_hood.sort_values("mean", ascending = False).round(0)
by_hood
```

**THE EXTREMES**

```python
print(by_hood.head(3))
print(by_hood.tail(3))
```

**COMPARING LOCATION AGAINST SIZE**

```python
#How strongly size relates to price
print(df["SalePrice"].corr(df["GrLivArea"]).round(2))
```

```python
#Price per square foot, by neighborhood
df["PricePerSqFt"] = df["SalePrice"] / df["GrLivArea"]
df.groupby("Neighborhood")["PricePerSqFt"].median().sort_values(ascending = False).round(0).head(5)

#If size explained everything, every neighborhood would pay
#about the same per sqare foot
```
